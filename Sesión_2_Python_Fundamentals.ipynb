Si la compra es menor a S/100.00. no hay descuento, si la compra está entre S/100 y S/299.99. recibe un 10% de descuento. Si la compra está entre S/500 o más, recibe un 20% de descuento. El programa debe mostrar: monto original, porcentaje de descuento, monto descontado y total a pagar y además debe pedir al usuario el monto de la compra.



```python
monto = float(input("Ingresa el monto de tu compra: "))

if monto <= 0:
  print("Error: el monto debe ser mayor que cero.")
else:
  if monto < 100:
    descuento = 0
  elif monto < 300:
    descuento = 10
  else:
    descuento = 20

  monto_descontado = monto * (descuento / 100)
  total_pagar = monto - monto_descontado

  print("\n--Resumen de la Compra--")
  print(f"Monto original: S/{monto:.2f}")
  print(f"Descuento: {descuento}%")
  print(f"Monto descontado: S/{monto_descontado:.2f}")
  print(f"Total a pagar: S/{total_pagar:.2f}")
```

**Procesamiento de datos y Built In**

Tienes una lista de precios de productos y necesitas calcular automáticamente el resumen del día.

*   Recorre cada precio de la lista con un bucle for
*   Sumar cada precio al total acumulado
*   Calcular el promedio dividiendo el total entre la cantidad de productos
*   Encontrar el precio más alto con max()
*   Encontrar el precio más bajo con min()
*   Mostrar todo formateado con 2 decimales



```python
precios = [25.50, 30.00, 15.75, 60.30, 8.90, 33.20]

total = 0
precio_max = precios[0]
precio_min = precios[0]

for precio in precios:
  total += precio

promedio = total / len(precios)

print("=== RESUMEN TOTAL DEL DÍA ===")
print(f"Total de Productos: {len(precios)} ")
print(f"Total Acumulado: S/{total:.2f}")
print(f"Promedio: S/{promedio:.2f} ")
print(f"Precio Máximo: S/{max(precios):.2f}")
print(f"Precio Mínimo: S/{min(precios):.2f}")

```

# **Sistema de notas del salón**

Desarrolla un sistema que gestione las notas de 4 estudiantes y genere un reporte completo del salón.
*   Pedir el nombre y nota de 4 estudiantes
*   Guardar los datos en una estructura (lista o diccionario)
*   Calcular el promedio del salón
*   Mostrar quién aprobó y quien desaprobó (aprobado > 11)
*   Mostrar la nota más alta y más baja



```python
estudiantes = []

for i in range(4):
  nombre = input(f"Ingresa el nombre del alumno {i + 1}: ")
  nota = float(input(f"Ingresa la nota de {nombre}: "))
  estudiantes.append({"nombre" : nombre, "nota": nota}1)
notas = [e["nota"] for e in estudiantes]
promedio = sum(notas) / len(notas)

aprobados = [e for e in estudiantes if e["nota"] > 11]
desaprobados = [e for e in estudiantes if e["nota"] <= 11]

print("\n===REPORTE DEL SALÓN=== ")
for e in estudiantes:
  print(f"{e["nombre"]} : {e["nota"]}")
print(f"\nPromedio del Salón: {promedio:.2f}")
print(f"Nota más alta: {max(notas)}")
print(f"Nota más baja: {min(notas)}")

print("\nAprobados: ")
if aprobados:
  for e in aprobados:
    print(f" {e["nombre"]} ({e["nota"]})")
else:
  print("No hay estudiantes aprobados.")


print("Desaprobados: ")
if desaprobados:
  for e in desaprobados:
    print(f" {e["nombre"]} ({e["nota"]})")
else:
  print("No hay estudiantes desaprobados.")
```
