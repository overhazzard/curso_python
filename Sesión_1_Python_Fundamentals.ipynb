# **OPERADORES, STRINGS, E INPUT OUTPUT:**


*   Ejercicio 1: Declara dos números enteros a elección y muestra su suma, resta, multiplicación, división, división entera y módulo. Etiqueta cada resultado.



```python
numero_1 = 28
numero_2 = 13
suma = numero_1 + numero_2
resta = numero_1 - numero_2
divison = numero_1 / numero_2
multiplicacion =  numero_1 * numero_2
divison_entera =  numero_1 // numero_2
modulo =  numero_1 % numero_2
print(f"La suma es: {suma}")
print(f"La resta es: {resta}")
print(f"La divison es: {divison}")
print(f"La multiplicacion es: {multiplicacion}")
print(f"La divison entera es: {divison_entera}")
print(f"El modulo es: {modulo}")
```

*   Pida el nombre del usuario y su saldo actual




*   Pida dos gastos realizados durante el día
* Calcule el total gastado y el saldo restante
* Calcule qué porcentaje del saldo se gastó
* Muestre un resumen final con todos los datos formateados



```python
nombre = input("Ingrese su nombre: ")
saldo_actual = float(input("Ingrese su saldo actual: "))

gasto_1 = float(input("Ingrese el primer gasto: "))
gasto_2 = float(input("Ingrese el segundo gasto: "))

total_gastado = gasto_1 + gasto_2

saldo_restante = saldo_actual - total_gastado

porcentaje_gastado = (total_gastado / saldo_actual) * 100

print(f"Hola {nombre}, tu saldo actual es de S/.{saldo_actual}.")
print(f"Durante el día tuviste 2 gastos: S/. {gasto_1} y S/.{gasto_2}")
print(f"El porcentaje que gastaste de tu saldo actual es el siguiente: {porcentaje_gastado}")
```
